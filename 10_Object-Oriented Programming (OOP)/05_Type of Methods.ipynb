# Instance Method 

```python
class Student:
    def __init__(self, name, cgpa):    # Constructor method (special method)
        self.name = name
        self.cgpa = cgpa

    def show_info(self):     # instance method
        print(f"Name: {self.name}, CGPA: {self.cgpa}")

s1 = Student("Raihan", 3.45)
s1.show_info()   # method call
```

```python
class Student:
    Unic_name="INTER"
    
    def __init__(self, name, cgpa):    # Constructor method (special method)
        self.name = name
        self.cgpa = cgpa

    def show_info(self):     # instance method
        print(f"Name: {self.name}, CGPA: {self.cgpa}",Student.Unic_name)
    @classmethod
    def up_Unic_name(cls,u_name):
        cls.Unic_name =u_name
      
    

s1 = Student("Raihan", 3.45)
s2=Student("Tasfis", 3.33)
s1.show_info()   # method call
s2.show_info()  
Student.up_Unic_name("NITER")
s1.show_info()   # method call
s2.show_info() 
```
