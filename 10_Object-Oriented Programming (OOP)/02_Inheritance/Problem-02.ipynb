```python
#now if we want to use charecteristies from Father class to children class then
#base class
class Father:
    def Father_name(self):
        print("Programming")
        
class Mother:
    def Mother_name(self):
        print("Design")

class children(Father):
    def output(self):
        print("Hello")
        
obj1=children()
obj1.output() 
obj1.Father_name()
```

# Diffrent Type of Inheritance

```python
#Single Inheritance

class Father:
    def Father_name(self):
        print("Programming")
        
class children(Father):
    def output(self):
        print("Hello")
        
obj1=children()
obj1.output() 
obj1.Father_name()
```

```python
#Multiple Inheritance

class Father:
    def Father_name(self):
        print("Programming")
        
class Mother:
    def Mother_name(self):
        print("Design")

class children(Father,Mother):
    def output(self):
        print("Hello")
        
obj1=children()
obj1.output() 
obj1.Father_name()
obj1.Mother_name()
```

```python
#Multi-level Inheritance

class Father:
    def Father_name(self):
        print("Programming")
        
class Mother(Father):
    def Mother_name(self):
        print("Design")

class children(Mother):
    def output(self):
        print("Hello")
        
obj1=children()
obj1.output() 
obj1.Mother_name()
obj1.Father_name()

```

```python
#Hierical Inheritance

class Father:
    def Father_name(self):
        print("Programming")
        
class Mother(Father):
    def Mother_name(self):
        print("Design")

class children(Father):
    def output(self):
        print("Hello")
        
obj1=children()
obj1.output() 
obj1.Father_name()
#obj1.Mother_name() ; Because children didnot inherite Mother class
```
