```python
class person:
    def __init__(self, name, age):
        self.name = name           # public variable
        self._age = age            # protected variable (by convention)
        self.__ID = '63'           # private variable

    def display(self):
        return self.__ID           # accessing private variable from inside class

P1 = person("Raihan", 25)
print(P1.display())                # prints: 63
```

# Class or static variable

```python
class Player:
    def __init__(self, run):
        self.run = run  # instance variable

    def hit_four(self):
        self.run += 4

    def hit_six(self):
        self.run += 6

sakib = Player(0)
tamim = Player(0)
sakib.hit_six()
tamim.hit_four()
tamim.hit_four()
print("Sakib:", sakib.run)
print("Tamim:", tamim.run)

print("Tamim", tamim.__dict__)
print("Sakib", sakib.__dict__)
```

```python
class Player:
    team_run=0       #static or class variable
    def __init__(self, run):
        self.run = run  # instance variable

    def hit_four(self):
        self.run += 4
        Player.team_run +=4

    def hit_six(self):
        self.run += 6
        Player.team_run +=6


sakib = Player(0)
tamim = Player(0)
sakib.hit_six()
sakib.hit_six()
tamim.hit_four()
tamim.hit_four()
print("Sakib:", sakib.run)
print("Tamim:", tamim.run)
print("Total team run:" ,Player.team_run)
print("Sakib er team run :", sakib.team_run)
print("Tamimer team run:", tamim.team_run)


print("Tamim", tamim.__dict__)
print("Sakib", sakib.__dict__)


#but print(Player.team_run)  output: 0 ; we didnot create any object
```
