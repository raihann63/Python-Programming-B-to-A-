# We can more simplyfy this code using Method and Constructors

# More Simplyfy using method

```python
#using method we can reuse and more small

class student:
    Id=""
    CGPA=""
    def result(own):
        print(f"Student Class ID : {own.Id}\nHis CGPA :{own.CGPA}")
        
    
Raihan=student()
Raihan.Id=2102063
Raihan.CGPA=3.45
Raihan.result()

print()  # <- This adds a blank line between the two outputs

SOhag=student()
SOhag.Id=2102016
SOhag.CGPA=3.49
SOhag.result()
```

```python
#using method we can reuse and more small

class student:
    Id=""
    CGPA=""
    
    '''এটাকে বলে Object Binding
       অর্থাৎ object নিজের ভিতরের property অ্যাক্সেস করার জন্য self/own দরকার হয়।'''
    
    def set_value(own,Id,CGPA): 
        own.ID=Id
        own.CGPA=CGPA
    def result(own):
        print(f"Student Class ID : {own.ID}\nHis CGPA :{own.CGPA}")
        
    
Raihan=student()
Raihan.set_value(63,3.45)
Raihan.result()

print()  # <- This adds a blank line between the two outputs

SOhag=student()
SOhag.set_value(16,3.49)
SOhag.result()
```

# More Simplyfy using constrator


```python
#constructor is spacial type of method

class student:

    '''এটাকে বলে Object Binding
       অর্থাৎ object নিজের ভিতরের property অ্যাক্সেস করার জন্য self/own দরকার হয়।'''
    
    def __init__(own,Id,CGPA):
        own.ID=Id
        own.CGPA=CGPA
    def result(own):
        print(f"Student Class ID : {own.ID}\nHis CGPA :{own.CGPA}")
        
    
Raihan=student(63,3.45)
Raihan.result()

print()  # <- This adds a blank line between the two outputs

SOhag=student(16,3.49)
SOhag.result()
```

# Default Constructor

```python
class Student:
    def __init__(self):
        self.name = "Unknown"
        self.cgpa = 0.0

    def show(self):
        print(f"Name: {self.name}, CGPA: {self.cgpa}")
        
s1 = Student()
s1.show()
```

# Parameterized Constructor

```python
class Student:
    def __init__(self, name, cgpa):
        self.name = name
        self.cgpa = cgpa

    def show(self):
        print(f"Name: {self.name}, CGPA: {self.cgpa}")

s1 = Student("Raihan", 3.45) # here s1 that we crete object
s1.show()
```

# How to work Instance Variable & Instance Method

# Instance Variable


```python
class Student:
    def __init__(self, name, cgpa):
        self.name = name          # instance variable
        self.cgpa = cgpa          # instance variable

s1 = Student("Raihan", 3.45)
s2 = Student("Sohag", 3.49)

print(f"Student 1 Name: {s1.name}, CGPA: {s1.cgpa}")
print(f"Student 2 Name: {s2.name}, CGPA: {s2.cgpa}")
```

# Instance Method 

```python
class Student:
    def __init__(self, name, cgpa):    # Constructor method (special method)
        self.name = name
        self.cgpa = cgpa

    def show_info(self):     # instance method
        print(f"Name: {self.name}, CGPA: {self.cgpa}")

s1 = Student("Raihan", 3.45)
s1.show_info()   # method call
```
