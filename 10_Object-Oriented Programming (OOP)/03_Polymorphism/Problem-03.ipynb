```python
class student:
    def __init__(self,age,name):
    #__init__ হলো constructor method → যেটা object তৈরি হলে অটোমেটিক কল হয়
        self.Age=age       #age → হলো constructor-এর argument, যেটা object তৈরি করার সময় আমরা পাঠাবো
                           #self.Age = age মানে: object-এর ভিতরে Age নামে একটা data রাখা হলো
        self.name=name
        
student1=student(25,"Raihan")
print(student1.Age,student1.name)

student2=student(55,"Rohan")
print(student2.Age,student2.name)
```

```python
#Method overloading

class calculator:
    def add(self,x,y=0,z=0):
        return x+y+z
    
cal1=calculator()
print(cal1.add(5))
print(cal1.add(5,8))    
print(cal1.add(6,9,3))


#note: if you seee carefully you can see,here we used diffrent argumnent in diffrent time.
```

```python
#Method Override

class student:
    def subject(self):
        print("My subject")
        
class Science(student):
    pass

student1=student()
student1.subject()


student2=student()
student2.subject()      



#Note: see same output beacuse we could inherate
```

```python
class student:
    def subject(self):
        print("My subject")
        
class Science(student):
    def subject(self):
        super().subject()
        
        '''এই subject() method Science class-এর ভিতরে আবার define করা হয়েছে, 
                        যদিও একই নামের method student class-এ আগেই ছিল।
                        Override মানে:Parent class-এর method কে child class-এ আবার define করা'''
        print("Only science")
   
student2=Science()
student2.subject()      
```
