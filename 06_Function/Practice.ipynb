# Step-01: No Parameter, No Return

```python
def fun():
    print("Ease function")
fun()
```

# Step-02: With Parameter, No Return

```python
def fun(x):
    print(x)
fun(2)
```

```python
def function(x,y,z):
    print(x+y+z)
function(1,2,3)
```

# Step-03: With Parameter, WIth Return

```python
def function(x,y,z):
    return(x+y+z)
result=function(1,2,3)
print(result)
```

```python
def test(x):
    return x*3
result=test(9)
print("You result=",result)
```

```python
def function(x):
    return x*3
x=int(input("Enter the number= "))
result=function(x)
print("You result= ",result)
```

# Arbitrary Argument

```python
#For unknown argument pass 
def fun (*x):
    print(x)
fun(6,4,5,7,8,8)
```

# Keyword Argument

```python
def fun(k1,k2,k3):
    print("flower: ",k1,k2,k3)
fun(k1="Rose",k2="Waterlily",k3="Matigold")
```

```python
def fun(**k1):
    print("flower: ",k1)
fun(k1="Rose",k2="Waterlily",k3="Matigold")
```

```python
def fun(**k1):
    print("flower: ",k1)
fun(Full="Rose",Food="Apple")
```
